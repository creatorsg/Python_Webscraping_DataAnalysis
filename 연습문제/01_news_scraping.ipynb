### 연습문제 1

아래의 url에서 뉴스기사의 링크와 제목을 출력하세요.

    url = 'https://news.daum.net/economy'


```python
import requests
import bs4
from bs4 import BeautifulSoup
```

```python
url = "https://news.daum.net/economy"
print(url)

req_header = {"user-agent": "Mozilla/5.0"}

res = requests.get(url, headers=req_header)
print(res.status_code)


if res.ok:
    res.encoding = "utf-8"

    html = res.text
    soup = BeautifulSoup(html, "html.parser")

    li_tags = soup.select("ul.list_newsheadline2 li")

    for li_tag in li_tags:
        a_tag = li_tag.find("a")

        if a_tag:
            link = a_tag.get("href")
            strong_tag = li_tag.select_one("strong.tit_txt")

            if strong_tag:
                title = strong_tag.text.strip()

                print(link)
                print(title)
                print()

else:
    print(f"Error Code = {res.status_code}")
    exit()
```

### 연습문제 2 - 1. Nate 뉴스기사 제목 스크래핑하기

### nata뉴스 기본정보

홈 : https://news.nate.com/ <br>
최신 뉴스 : https://news.nate.com/recent?mid=n0100 <br>
정치 : https://news.nate.com/section?mid=n0200 <br>
경제 : https://news.nate.com/section?mid=n0300 <br>
사회 : https://news.nate.com/section?mid=n0400 <br>
세계 : https://news.nate.com/section?mid=n0500 <br>
IT/과학 : https://news.nate.com/section?mid=n0600 <br>

<p> 즉, 홈을 제외한 나머지는 https://news.nate.com/section?mid=n + {고유 번호}의 규칙을 고수하고 있다. </p>

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from IPython.display import Image, display
```

```python
nate_config = {
    '최신뉴스': 'n0100',
    '정치': 'n0200',
    '경제': 'n0300',
    '사회': 'n0400',
    '세계': 'n0500',
    'IT/과학': 'n0600'
}

nate_site = 'https://news.nate.com/section?mid='
```

```python
def verify_news(section_name):
    section = nate_config.get(section_name)

    if section is None:
        print(f"'{section_name}'은(는) 지원하지 않는 섹션입니다.")
        return None

    if section_name == '최신뉴스':
        return 'https://news.nate.com/recent?mid=' + section

    return nate_site + section


def print_one_news(title, link, img_tag, base_url):
    print(title)
    print(link)

    if img_tag:
        src = img_tag.get('src')
        img_url = urljoin(base_url, src)
        display(Image(url=img_url))
    else:
        print('(이미지 없음)')

    print()

def print_news_with_image(section_name):
    url = verify_news(section_name)
    if url is None:
        return

    print(f'======> {url} {section_name} 뉴스 <======')

    req_header = {'user-agent': 'Mozilla/5.0'}
    res = requests.get(url, headers=req_header)

    if not res.ok:
        print(f'Error Code = {res.status_code}')
        return

    res.encoding = res.apparent_encoding
    soup = BeautifulSoup(res.text, 'html.parser')

    top = soup.select_one('div.mlt01')
    if top:
        a_tag = top.select_one('a[href*="/view/"]')
        if a_tag:
            title = a_tag.get_text(strip=True)
            link = urljoin(url, a_tag.get('href'))
            img_tag = top.find('img')
            print_one_news(title, link, img_tag, url)

    li_tags = soup.select('ul.mduList1 li')

    for li_tag in li_tags:
        h2 = li_tag.find('h2')
        if not h2:
            continue

        a_tag = h2.find('a')
        if not a_tag:
            continue

        title = a_tag.get_text(strip=True)
        link = urljoin(url, a_tag.get('href'))
        img_tag = li_tag.find('img') 

        print_one_news(title, link, img_tag, url)
```

```python
print_news_with_image('경제')
```
