### 네이버 책 검색 API 호출하기 (필수)

```python
import requests
import json
import os
import pandas as pd
from dotenv import load_dotenv
 
load_dotenv()
 
CLIENT_ID = os.getenv("CLIENT_ID")
CLIENT_SECRET = os.getenv("CLIENT_SECRET")
 
headers = {
    'X-Naver-Client-Id': CLIENT_ID,
    'X-Naver-Client-Secret': CLIENT_SECRET
}
```

```python
def search_books(query):
    payload = {
        'query': query,
        'display': 100,
        'sort': 'sim'
    }
    
    url = 'https://openapi.naver.com/v1/search/book.json'

    res = requests.get(url, params=payload, headers=headers)
 
    if not res.ok:
        print(f'API 요청 실패 (status={res.status_code})')
        print(res.text)
        return []
 
    return res.json()['items']

def save_books_json(items, path='data/books.json'):
    dir_name = os.path.dirname(path)
    if dir_name:
        os.makedirs(dir_name, exist_ok=True)
 
    with open(path, 'w', encoding='utf-8') as file:
        json.dump(items, file, ensure_ascii=False, indent=2)
 
    print(f'{len(items)}건을 {path}에 저장했습니다.')
 
 
def load_books_df(path='data/books.json'):
    with open(path, 'r', encoding='utf-8') as file:
        items = json.load(file)
 
    return pd.DataFrame(items)
```

# 1. 질문 :  검색어로  찾은  책 목록을 json 파일로 저장하기

```python
books = search_books('파이썬')
save_books_json(books)
```
