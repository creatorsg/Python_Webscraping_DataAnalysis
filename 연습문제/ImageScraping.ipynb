### 2-2. 하나의 네이버 웹툰과 1개의 회차에 대한 Image 다운로드 하기

```python
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse
```

```python
def get_episode_soup(url):
    res = requests.get(url) 

    if not res.ok:
        print(f'페이지 요청 실패 (status={res.status_code})')
        return None

    return BeautifulSoup(res.text, 'html.parser')


def extract_image_urls(soup):
    img_tags = soup.select("img[src*='IMAG01']")
    return [img_tag['src'] for img_tag in img_tags]


def make_episode_dir(title, no):
    dir_name = os.path.join('img', title, str(no))
    os.makedirs(dir_name, exist_ok=True)
    return dir_name


def download_image(img_url, dir_path, referer):
    req_header = {'referer': referer}
    res = requests.get(img_url, headers=req_header)

    if not res.ok:
        print(f'다운로드 실패 (status={res.status_code}): {img_url}')
        return False

    file_name = os.path.basename(urlparse(img_url).path)
    file_path = os.path.join(dir_path, file_name)

    with open(file_path, 'wb') as file:
        file.write(res.content)

    print(f'{file_name} ({len(res.content)} bytes) 저장 완료')
    return True
```

```python
def download_one_episode(title, no, url):
    soup = get_episode_soup(url)
    if soup is None:
        return

    img_url_list = extract_image_urls(soup)
    print(f'이미지 {len(img_url_list)}개 발견')

    if not img_url_list:
        print('이미지를 찾을 수 없습니다.')
        return

    dir_path = make_episode_dir(title, no)
    print(f'저장 경로: {dir_path}')

    success_count = 0
    for idx, img_url in enumerate(img_url_list, 1):
        print(f'[{idx}/{len(img_url_list)}]', end=' ')
        if download_image(img_url, dir_path, referer=url):
            success_count += 1

    print(f'완료: {success_count}/{len(img_url_list)}장 저장')
```

```python
download_one_episode('일렉시드', 341, 'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed')
```
